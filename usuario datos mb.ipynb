```python
from math import *
import numpy as np
```

```python
#El ususario ingresa los datos
funcion=input("Ingresa la función f(x): ") #Ejemplo: x**2+10-4
```

```python
a=float(input("Ingresa el valor de a: ")) #Quedando el intervalo como [a,b]
```

```python
b=float(input("Ingresa el valor de b: ")) #Quedando el intervalo como [a,b]
```

```python
tol=float(input("Ingresa la tolerancia: ")) #Hasta que decimal llegar, ejemplo: 0.000001
```

```python
#Definimos la función que ingresó el usuario
def f(x):
    return eval(funcion)
```

```python
#Guardando los valores iniciales
a0=a
b0=b

nmax=100 #Número máximo de iteraciones
error=100 #Error inicial
niter=0 #Número de iteraciones
```

```python
# Método de Bisección

#Evaluación del primer punto medio
m=a+(b-a)/2

#Evaluación de la función en los puntos
fa=f(a)
fb=f(b)
fm=f(m)

print("\n# iter\t\t a \t\t f(a) \t\t b \t\t f(b) \t\t m \t\t f(m) \t\t error")

print("{0} \t\t {1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter, a, fa, b, fb, m, fm, error))

# Ciclo iterativo
while error > tol and niter < nmax:

    m=a+(b-a)/2
    fm=f(m)

    if np.sign(fa) == np.sign(fm):
        a=m
        fa=f(a)
    else:
        b=m
        fb=f(b)

    m=a+(b-a)/2
    fm=f(m)

    error=abs(b-a)

    niter=niter+1

    print("{0}\t {1:8.6f}\t {2:8.6f}\t {3:8.6f}\t {4:8.6f}\t {5:8.6f}\t {6:8.6f}\t {7:8.6f}".format(niter, a, fa, b, fb, m, fm, error))
    

print("\nLa raíz de la función dada en el intervalo [{0:.4f}, {1:.4f}] es: {2:.7f}".format(a0, b0, m))

print("Número de iteraciones:", niter)
print("Error final:", error)
```
