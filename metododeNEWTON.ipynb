```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def f(x): #Definimos la función
 return np.cos(x)-x #La función es:
```

```python
def df(x): #Definimos la derivada
    return -np.sin(x)-1 #La derivada es:
```

```python
xx = np.linspace(0, 1, 100)
plt.plot(xx, f(xx), label="f(x)")
plt.plot(xx, df(xx), label="f'(x)")
plt.axhline(y=0, color="black")
plt.axvline(x=0, color="black")
```

```python
#Datos iniciales
p0=0.5
tol=0.000001
N0=100
error=100
i=1
```

```python
lista=[]

while i<=N0 and error>tol:
    if df(p0)==0:
        print("Error: la derivada es cero.")
        break

    p1=p0-f(p0)/df(p0)
    error=np.abs(p1-p0)

    lista.append([i, p0, f(p0), p1, f(p1), error])

    p0=p1
    i=i+1
```

```python
tabla=pd.DataFrame(lista, columns=["Iteracion", "p0", "f(p0)","p1", "f(p1)", "Error"])
display(tabla)
```

```python
if error<=tol:
    print("El procedimiento fue exitoso.")
    print("La raiz aproximada es:", p0)
    print("Numero de iteraciones:", i-1)
else:
    print("El metodo fallo o no alcanzo la tolerancia.")
```
