```python
import numpy as np
import pandas as pd
import sympy as sp
```

```python
x=sp.Symbol('x')
xi=np.array([1.0,1.3,1.6,1.9,2.2])
fi=np.array([0.7651977,0.6200860,0.4554022,0.2818186,0.1103623])
valor=1.5
n=len(xi)
Q=np.full((n, n), np.nan)
Q[:,0]=fi
```

```python
#Metodo de Neville
for i in range(1, n):
    for j in range(1, i + 1):

        Q[i,j]=((valor-xi[i-j])*Q[i,j-1]-(valor-xi[i])*Q[i-1,j-1])/(xi[i]-xi[i-j])
```

```python
tabla=pd.DataFrame(Q)

tabla.index=[f"Q{i}" for i in range(n)]
tabla.columns=[f"Columna{j}" for j in range(n)]

print("TABLA DE INTERPOLACION DE NEVILLE")
display(tabla.round(7))
tabla=pd.DataFrame(Q)

tabla.index=[f"Q{i}" for i in range(n)]
tabla.columns=[f"Columna{j}"for j in range(n)]

print("TABLA DE INTERPOLACION DE NEVILLE")
display(tabla.round(7))
```

```python
resultado=Q[n-1, n-1]

print("Aproximacion de f(1.5):")
print(resultado)
```

```python
# Polinomio de Lagrange
P=0

for i in range(n):
    termino=1

    for j in range(n):
        if i !=j:
            termino*=(x-sp.Float(xi[j]))/(sp.Float(xi[i])-sp.Float(xi[j]))

    P+=sp.Float(fi[i])*termino

P=sp.expand(P)

print("POLINOMIO DE INTERPOLACION:")
sp.pprint(P)

print("Evaluacion del polinomio en x=1.5:")
print(sp.N(P.subs(x,valor),10))
```
