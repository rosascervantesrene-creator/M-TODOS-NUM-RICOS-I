```python
import numpy as np
import pandas as pd
import sympy as sp
```

```python
x=sp.Symbol('x')
xi=[1.0,1.3,1.6,1.9,2.2]
fi=[0.7651977,0.6200860,0.4554022,0.2818186,0.1103623]
n=len(xi)
F=sp.zeros(n, n)
```

```python
#Valores de la funcion
for i in range(n):
    F[i,0]=sp.Rational(str(fi[i]))
```

```python
#Diferencias divididas
for j in range(1,n):
    for i in range(j,n):
        F[i,j]=(F[i,j-1]-F[i-1,j-1])/(sp.Rational(str(xi[i]))-sp.Rational(str(xi[i-j])))
```

```python
tabla=pd.DataFrame(np.array(F.tolist(),dtype=float))

tabla.index=[f"i={i}" for i in range(n)]
tabla.columns=["f[xi]","1ra diferencia","2da diferencia","3ra diferencia","4ta diferencia"]

print("TABLA DE DIFERENCIAS DIVIDIDAS")
display(tabla.round(7))
```

```python
coeficientes=[F[i,i] for i in range(n)]

print("\nCOEFICIENTES DE NEWTON")

for i, c in enumerate(coeficientes):
    print(f"F{i},{i}=",sp.N(c, 10))
```

```python
P=coeficientes[0]

for i in range(1, n):
    termino=1

    for j in range(i):
        termino*=(x - sp.Rational(str(xi[j])))

    P+=coeficientes[i]*termino
```

```python
P=sp.expand(P)

print("\nPOLINOMIO DE NEWTON:")
sp.pprint(P)
```

```python
sp.plot(P, (x,1.0,2.2),title="Interpolacion de Newton",xlabel="x",ylabel="P(x)")
```
