```python
Queremos resolver la ecuación $x^3-4x^2-10=0$ en el intervalo $[1,2]$

Para aplicar el **método de bisección** verificamos que la función $f(x)$ está evaluada en los extremos opuestos tiene signos opuestos del intervalo $[a,b]$ tiene signos opuestos, es decir, $f(a)\times f(a)<0$

El teorema del valor intermedio nos asegura que existe un número $p\in [a,b]$ tal que es una raíz de $f(x)$, es decir, $f(p)=0$

Aproximaremos la solución, calculando de manera iterativa el punto medio del intervalo: $$p_i=\frac{(b_i-a_i)}{2}$$
```

```python
 from math import *
import numpy as np
import matplotlib.pyplot as plt
```

```python
Definimos la función $f(x)$ asociada a la ecuación $f(x)=0$
```

```python
def f(x):
 return x**3+4*x**2-10.0
```

```python
xx=np.linspace(0,3) #genera una reticula o una partición
plt.plot(xx,(f(xx)),)
plt.title("Gráfica $f(x)=x^3+4x^2-10.0$")
plt.xlabel("$x$")
plt.ylabel("$f(x)$")
# para hacer un punto plt.scatter(
plt.axhline(y=0,color='red') #trazar linea horizontal
```

```python
#ingreso datos de entrada para los diferentes métodos a trabajar
a=1
b=2

#guardar valores iniciales
a0=a
b0=b

#guardar valores iniciales del error y el número de iteraciones
tol=0.000001
nmax=100 #iteraciones
error=100
niter=0
```

```python
# Método de Bisección

# evaluo primer valor medio
m = a + (b - a)/2

#Evaluacion de la función en los puntos a, b y m
fa = f(a)
fb = f(b)
fm = f(m)

print("# iter\t\t a \t\t f(a) \t\t b \t\t f(b) \t\t m \t\t f(m) \t\t error")
print("{0} \t\t {1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter, a0, fa, b0, fb, m, fm, error ))

# ciclo iterativo
while error > tol and niter < nmax:
    m = a + (b - a) / 2
    if np.sign(fa) == np.sign(fm):
        a = m
        fa = f(a)
    else:
        b = m
        fb = f(b)

    m = a + (b - a)/2
    fm = f(m)
    error = abs(b - a)
    niter += 1
    print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f} \t {4:6.6f} \t {5:6.6f} \t {6:6.6f} \t {7:6.6f}".format(niter, a, fa, b, fb, m, fm, error ))

print("La raíz de la función dada en el intervalo [{0:6.4f},{1:6.4f}] es {2:6.7f}".format(a0,b0,m))
```
